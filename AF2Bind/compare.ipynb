```python
import os, csv

for file in os.listdir('/home/unix/vkrhk/g2p-predictors/AF2Bind/output'):
    my_output = {}
    with open('/home/unix/vkrhk/g2p-predictors/AF2Bind/output/' + file) as f:
        reader = csv.reader(f, delimiter='\t')
        next(reader) # skip header
        for row in reader:
            my_output[row[0]] = float(row[3])

    javier_output = {}
    with open('/home/unix/vkrhk/g2p-predictors/AF2Bind/javier-output/' + file[:-3] + 'csv') as f:
        reader = csv.reader(f, delimiter=',')
        next(reader) # skip header
        for row in reader:
            javier_output[row[1]] = float(row[3])
    
    largest_difference = 0
    for key in my_output:
        difference = abs(my_output[key] - javier_output[key])
        if difference > largest_difference:
            largest_difference = difference

    af2bind_output = {}
    uniprot_id = file.split('-')[1]
    with open('/home/unix/vkrhk/g2p-predictors/AF2Bind/af2bind-output/' + uniprot_id + '_AF2BIND_predictions.csv', 'r') as f:
        reader = csv.reader(f, delimiter=',')
        next(reader) # skip header
        for row in reader:
            af2bind_output[row[1]] = float(row[2])
    
    largest_difference_af2bind = 0
    for key in my_output:
        if key not in af2bind_output:
            continue
        difference = abs(my_output[key] - af2bind_output[key])
        if difference > largest_difference_af2bind:
            largest_difference_af2bind = difference

    print(f'{file}, max diff. with AF2Bind: {largest_difference_af2bind}, max diff. with Javier: {largest_difference}')
## TODO: rerun the docker build!!
```
